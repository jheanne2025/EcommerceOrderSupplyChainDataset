```python
# Sostituisco i valori nulli con DEFAULT_DATETIME
# Colonne con valori nulli e il loro conteggio:
# order_delivered_timestamp    1889
# order_approved_at               9

# Load CSV / Carica CSV
dforderscopy = pd.read_csv("df_Orders.csv")

# Define default placeholder datetime / Definisco il segnaposto data/ora predefinito
DEFAULT_DATETIME = pd.Timestamp("1970-01-01 00:00:00")

# Replace empty / missing values in datetime columns / Sostituisco i valori vuoti/mancanti nelle colonne datetime
dforderscopy["order_delivered_timestamp"] = (
    pd.to_datetime(dforderscopy["order_delivered_timestamp"], errors="coerce")
    .fillna(DEFAULT_DATETIME)
)

dforderscopy["order_approved_at"] = (
    pd.to_datetime(dforderscopy["order_approved_at"], errors="coerce")
    .fillna(DEFAULT_DATETIME)
)

# Conversione coretta dei tipi dei dati
dforderscopy['order_purchase_timestamp'] = pd.to_datetime(dforderscopy['order_purchase_timestamp'])
dforderscopy['order_estimated_delivery_date'] = pd.to_datetime(dforderscopy['order_estimated_delivery_date'])

display(dforderscopy.info())

# Creo nuovo CSV file
dforderscopy.to_csv('dforderscopy.csv', index=False)
print("\nFile 'dforderscopy.csv' creato con successo!")
```
